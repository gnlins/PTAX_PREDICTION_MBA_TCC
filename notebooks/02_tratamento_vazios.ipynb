# Tratando os dados Vazios

## FORWARD FILL DAS VARIÁVEIS MACROECONÔMICAS

```python
variaveis_macro = [
    "Selic",
    "IPCA",
    "PIB",
    "Exportacoes",
    "Importacoes",
    "FEDFUNDS"
]

variaveis_financeiras = [
    "IBOV",
    "SP500",
    "Petroleo"
]
```

```python
df[variaveis_macro] = df[variaveis_macro].ffill()
```

## REMOVER LINHAS SEM REGISTRO DE USD/BRL

```python
df = df.dropna(subset=["USD/BRL"])
```

## INTERPOLAÇÃO DAS VARIÁVEIS FINANCEIRAS

```python
df[variaveis_financeiras] = (
    df[variaveis_financeiras]
    .interpolate(method="linear")
)
```

## VERIFICAR VALORES NULOS

```python
print("Valores nulos após tratamento:")
print(df.isna().sum())


print("\nPercentual de valores nulos:")
print((df.isna().mean() * 100).round(2))
```
